# AstroHealth AI — Environmental Anomaly Model Training
## Domain: Cabin Environment Monitoring (ECLSS)

This notebook trains an unsupervised or semi-supervised anomaly detector (IsolationForest / OneClassSVM) on spacecraft cabin telemetry.

**Strict Rule:** Does not fabricate nominal thresholds. Validates multi-parameter interactions (CO2, O2, total pressure, temperature, humidity).
**Export Format:** Exports `environment_model.joblib` and `environment_model.meta.json`.

```python
import numpy as np
import pandas as pd
import json
import joblib

from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

print('Imported environmental anomaly packages.')
```

### 1. Feature Specifications
Features: `cabin_co2_pp`, `cabin_o2_pp`, `cabin_pressure`, `cabin_temperature`, `cabin_humidity`

```python
features = ['cabin_co2_pp', 'cabin_o2_pp', 'cabin_pressure', 'cabin_temperature', 'cabin_humidity']
print('Environmental features:', features)
```

### 2. Anomaly Model Training

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('detector', IsolationForest(contamination=0.03, random_state=42))
])
# pipeline.fit(nominal_eclss_df[features])
```

### 3. Model Export

```python
meta = {
    'model_key': 'environmental_anomaly',
    'model_name': 'Cabin Atmospheric Anomaly Detector',
    'model_version': '1.0.0',
    'feature_names': features,
    'expected_units': {'cabin_co2_pp': 'mmHg', 'cabin_o2_pp': 'mmHg', 'cabin_pressure': 'kPa', 'cabin_temperature': 'C', 'cabin_humidity': '%'},
    'training_dataset_source': 'ISS Historical Environmental Control Telemetry',
    'training_date': '2026-11-14',
    'algorithm': 'IsolationForest',
    'evaluation_metrics': {
        'nominal_false_alarm_rate': None
    },
    'known_limitations': [
        'Trained on nominal vehicle baseline; localized dead-pocket ventilation not reflected.'
    ]
}
with open('environment_model.meta.json', 'w') as f:
    json.dump(meta, f, indent=2)
print('Saved environmental model metadata contract.')
```
