# AstroHealth AI — Behavioral Fatigue & Sleep Pattern Model Training
## Domain: Behavioral and Mental Health Monitoring

This notebook trains an offline machine learning model to estimate performance lapse likelihood from sleep duration actigraphy and PVT metrics.

**Strict Rule:** Splits data by astronaut subject ID to prevent information leakage across repeated mission days.
**Export Format:** Exports `behavioral_model.joblib` and companion `behavioral_model.meta.json`.

```python
import numpy as np
import pandas as pd
import json
import joblib

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

print('Imported behavioral analysis packages.')
```

### 1. Data Loading and Feature Inspection
Features: `sleep_duration_hours`, `pvt_mean_rt`, `pvt_lapses`
Target: `elevated_fatigue_risk`

```python
features = ['sleep_duration_hours', 'pvt_mean_rt', 'pvt_lapses']
target = 'elevated_fatigue_risk'
print('Features:', features)
```

### 2. Pipeline Training & Validation

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('rf', RandomForestClassifier(n_estimators=100, max_depth=4, random_state=42))
])
# pipeline.fit(X_train, y_train)
```

### 3. Model Export

```python
meta = {
    'model_key': 'behavioral_fatigue',
    'model_name': 'Cognitive Fatigue and Sleep Disruption Predictor',
    'model_version': '1.0.0',
    'feature_names': features,
    'expected_units': {'sleep_duration_hours': 'hours', 'pvt_mean_rt': 'ms', 'pvt_lapses': 'count'},
    'training_dataset_source': 'NASA Reaction Self-Test ISS Experiment',
    'training_date': '2026-11-14',
    'algorithm': 'RandomForestClassifier',
    'evaluation_metrics': {
        'test_f1_score': None
    },
    'known_limitations': [
        'Subjective fatigue self-reporting varies with workload.',
        'Requires adherence to daily PVT test schedules.'
    ]
}
with open('behavioral_model.meta.json', 'w') as f:
    json.dump(meta, f, indent=2)
print('Saved behavioral model metadata contract.')
```
